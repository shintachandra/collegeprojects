# Penjelasan Dataset
Dataset **food waste** ini berisi estimasi jumlah limbah makanan per kapita dan total per tahun di berbagai negara, mencakup sektor rumah tangga, ritel, dan layanan makanan.

Data ini memberikan gambaran tentang besarnya limbah makanan yang dihasilkan di tiap negara, termasuk distribusinya di berbagai sektor ekonomi. Selain itu, terdapat informasi mengenai tingkat kepercayaan terhadap estimasi yang diberikan, serta kode numerik internasional (M49) untuk mengidentifikasi negara dan wilayah geografis yang terkait.

Dataset ini bermanfaat untuk menganalisis perbedaan tingkat limbah makanan antar negara dan kawasan, serta membantu merancang kebijakan atau inisiatif pengurangan limbah di sektor-sektor kunci.

Berikut kolom-kolom yang ada pada dataset food waste:
- **Country**  
  Nama negara tempat data limbah makanan dikumpulkan.
- **Combined figures (kg/capita/year)**  
  Jumlah total limbah makanan per kapita per tahun dari semua sektor (rumah tangga, ritel, layanan makanan) dalam kilogram.
- **Household estimate (kg/capita/year)**  
  Estimasi limbah makanan yang dihasilkan oleh rumah tangga per kapita per tahun dalam kilogram.
- **Household estimate (tonnes/year)**  
Estimasi total limbah makanan yang dihasilkan oleh rumah tangga dalam suatu negara per tahun dalam ton.
- **Retail estimate (kg/capita/year)**  
Estimasi limbah makanan yang dihasilkan oleh sektor ritel per kapita per tahun dalam kilogram.
- **Retail estimate (tonnes/year)**  
Estimasi total limbah makanan yang dihasilkan oleh sektor ritel dalam suatu negara per tahun dalam ton.
- **Food service estimate (kg/capita/year)**  
Estimasi limbah makanan yang dihasilkan oleh sektor layanan makanan per kapita per tahun dalam kilogram.
- **Food service estimate (tonnes/year)**  
Estimasi total limbah makanan yang dihasilkan oleh sektor layanan makanan dalam suatu negara per tahun dalam ton.
- **Confidence in estimate**  
Tingkat kepercayaan terhadap estimasi data, yang mungkin diukur dalam kategori atau persentase.
- **M49 code**  
Kode numerik yang digunakan oleh Perserikatan Bangsa-Bangsa (PBB) untuk mengidentifikasi negara dan wilayah geografis
- **Region**  
Kawasan atau wilayah geografis dari negara yang terkait, mungkin digunakan untuk pengelompokan regional.
- **Source**  
Sumber data atau referensi yang digunakan untuk mengestimasi jumlah limbah makanan dalam dataset tersebut.

# Visualisasi

##**Heatmap korelasi pearson antar variabel numerik.**

```python
# Pilih hanya kolom numerik
df_numerik = df.select_dtypes(include=['float64', 'int64'])

# Hitung matriks korelasi menggunakan Pearson
korelasi = df_numerik.corr(method='pearson')

# Plot heatmap menggunakan seaborn
plt.figure(figsize=(10, 8))  # Ukuran heatmap
sns.heatmap(korelasi, annot=True, cmap='coolwarm', vmin=-1, vmax=1, cbar=True)

# Tampilkan plot
plt.title('Heatmap Korelasi Antar Variabel Numerik')
plt.show()
```

- dari sini kita dapat melihat ada beberapa korelasi yang sangat kuat dari variabel numerik yang ditandai dengan warna merah.  
- contoh:  
variabel retail estimate (tonnes/year) dengan variabel household estimate (tones/year) memiliki korelasi sangat tinggi sebesar 0.96.
- contoh lain:  
variabel retail estimate (tonnes/year) dengan variabel food service estimate (tonnes/year) juga memiliki korelasi yang sangat tinggi sebesar 0.96.

##**Pie Chart Tonnes Berdasarkan Sumbernya**

```python
import matplotlib.pyplot as plt

# Menghitung total tonase dari setiap kategori
total_household = df['Household estimate (tonnes/year)'].sum()
total_retail = df['Retail estimate (tonnes/year)'].sum()
total_food_service = df['Food service estimate (tonnes/year)'].sum()

# Menyimpan hasil dalam list
labels = ['Household', 'Retail', 'Food Service']
sizes = [total_household, total_retail, total_food_service]

# Membuat pie chart
plt.figure(figsize=(7, 7))
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90, colors=['#ff9999','#66b3ff','#99ff99'])
plt.title('Distribusi Tonnes Berdasarkan Sumbernya')
plt.axis('equal')  # Equal aspect ratio ensures that pie chart is circular.
plt.show()
```

- Sampah makanan terbanyak berasal dari household, diikuti dengan food service kemudian retail.

##**Bar Chart**

```python
# Get top 10 and bottom 10 countries
top_10 = df.nlargest(10, 'combined figures (kg/capita/year)')
bottom_10 = df.nsmallest(10, 'combined figures (kg/capita/year)')

# Combine the data
combined_df = pd.concat([top_10, bottom_10])

# Create a bar plot
plt.figure(figsize=(12, 8))
bars = plt.bar(combined_df['Country'], combined_df['combined figures (kg/capita/year)'], color=['red'] * 10 + ['green'] * 10)

# Add labels and title
plt.xticks(rotation=45)
plt.xlabel('Countries')
plt.ylabel('Combined figures (kg/capita/year)')
plt.title('Top 10 and Bottom 10 Countries by Food Waste (kg/capita/year)')

# Add a legend
red_patch = plt.Line2D([0], [0], color='red', lw=4, label='Top 10 (red)')
green_patch = plt.Line2D([0], [0], color='green', lw=4, label='Bottom 10 (green)')
plt.legend(handles=[red_patch, green_patch], loc='best')

# Show the plot
plt.tight_layout()
plt.show()
```

- dari bar chart di atas kita dapat melihat 10 negara dengan food waste terbanyak (ditandai warna merah) dan 10 negara dengan food waste paling sedikit (ditandai warna hijau).

```python
# Mengambil subset DataFrame dan menghitung rata-rata berdasarkan region
df_region = df[["Region", "Household estimate (kg/capita/year)", "Food service estimate (kg/capita/year)", "Retail estimate (kg/capita/year)", "combined figures (kg/capita/year)"]]
df_region = df_region.groupby(["Region"], as_index=False).mean()
df_region = df_region.sort_values(by=['combined figures (kg/capita/year)'], ascending=False)

# Menghapus kolom combined figures
df_region.pop("combined figures (kg/capita/year)")

# Membuat plot dengan ukuran lebih besar
ax = df_region.plot(x='Region', kind='bar', legend=True, stacked=True, title='Region-wise Aggregate Food Waste Per-Capita (kg/year)', figsize=(12, 8))

# Menambahkan legend
plt.legend(["Household", "Food Service", "Retail"])
plt.xlabel('Region')
plt.ylabel('Estimated Food Waste (kg/capita/year)')

# Menampilkan plot
plt.tight_layout()
plt.show()
```

- dari plot di atas kita dapat melihat bahwa di 17 region kebanyakan food waste dihasilkan dari limbah rumah tangga, lalu food service kemudian yang paling sedikit adalah industri ritel

## Choropleth

```python
df
```

```python
# Assuming df is your DataFrame with food waste data
food_waste_data = df

# Create the choropleth map
fig = px.choropleth(food_waste_data,
                    locations='Country',  # Country names
                    locationmode='country names',  # Use country names for locations
                    color='combined figures (kg/capita/year)',  # Values to color by
                    color_continuous_scale='Blues',  # A more appealing color scale
                    labels={'combined figures (kg/capita/year)': 'Combined Figures (kg/capita/year)'},
                    title='Food Waste in Asia (kg/capita/year)',
                    hover_name='Country',  # Display country names on hover
                    hover_data=['combined figures (kg/capita/year)']  # Show food waste data on hover
                    )

# Update layout to focus on Asia but show all countries
fig.update_geos(
    projection_type="mercator",  # Type of projection
    visible=True,  # Show all countries
    fitbounds="locations",  # Automatically fit bounds to the specified locations
    showcoastlines=True,  # Show coastlines
    coastlinecolor="Black",  # Color of the coastlines
    showland=True,  # Show land
    landcolor="lightgray",  # Color of land
    subunitcolor="Black",  # Color of country borders
)

# Update layout with increased size
fig.update_layout(
    title_font=dict(size=24, family='Arial', color='darkblue'),
    geo=dict(
        bgcolor='lightblue',  # Background color of the map
        landcolor='lightgray',  # Land color
        subunitcolor='gray',  # Color for subunits (countries)
        countrycolor='gray'  # Color for countries
    ),
    width=1000,  # Increase width of the figure
    height=600   # Increase height of the figure
)

# Show the figure
fig.show()
```

Choropleth menunjukan Combined figures (kg/capita/year) untuk setiap negara
